```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv(r'C:\Users\riazu\Desktop\Data-Science\Machine-Learning\2-Linear-Regression\placement.csv')
data.head(3)
```

```python
sns.scatterplot(x='cgpa', y='package', data=data)
plt.show()
```

```python
data.isnull().sum()
```

```python
# Create input/independent and output/dependent variables
x = data[['cgpa']]
y = data['package']
```

```python
# split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=35)
```

```python
# import the LinearRegression class from sklearn.linear_model
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
# fit the model to the training data
lr.fit(x_train, y_train)
```

```python
# y = m*x+c ---> m is the slope/coefficient and c is the intercept
print('Slope/Coefficient:', lr.coef_)
```

```python
# y = m*x+c ---> m is the slope/coefficient and c is the intercept
print('Intercept:', lr.intercept_)
```

```python
# this becomes the actual equation of the line of best fit for the data points in the scatter plot
# y = 0.54545922*x - -0.8123564468123363
```

```python
# Testing accuracy of the model
lr.score(x_test, y_test)
```

```python
lr.predict([[6.89]])
```

```python
0.54545922*6.89 - 0.8123564468123363
```

```python
y_pred = lr.predict(x)
```

```python
sns.scatterplot(x='cgpa', y='package', data=data)
plt.plot(data['cgpa'], y_pred, color='red')
plt.legend(['Data Points', 'Regression Prediction Line'])
plt.show()
```

```python
# Test the model with a test split and check the accuracy of the model
lr.predict(x_test)
```
