## Logistic Regression - Binomial Classification (Multiple Inputs)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('placement-dataset.csv')
df.head(5)
```

```python
# first, check whether the data is linearly separable (applicable for logistic regression)
# hue='placement' classifies the data based on placement column
sns.scatterplot(x='cgpa', y='iq', data=df, hue='placement') 
plt.show()
```

```python
# Create input and output variables
x = df.iloc[:,:-1]
y = df['placement']
```

```python
from sklearn.model_selection import train_test_split
```

```python
# split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
# initialize and train Logistic Regression
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
# Accuracy
lr.score(x_test, y_test)
```

```python
# predict where a student with cgpa = 6.8 and iq = 123 will be placed or not
lr.predict([[6.8, 123]])
```

```python
# This library is neded for plotting decision boundaries
from mlxtend.plotting import plot_decision_regions
```

```python
# Visualize decision boundaries
# input and output variables are converted into arrays bcz this library needs data in array form
plot_decision_regions(X=x.to_numpy(), y=y.to_numpy(), clf=lr, colors='blue,green')
plt.show()
```

```python
# coefficients
lr.coef_
```

```python
# intercept/constant
lr.intercept_
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
# confusion matrix
cm = confusion_matrix(y_test, lr.predict(x_test))
cm
```

```python
# Visualize confusion matrix
sns.heatmap(cm, annot=True)
plt.show()
```

```python
# calculate precision score
precision_score(y_test, lr.predict(x_test))*100
```

```python
# calculate recall score
recall_score(y_test, lr.predict(x_test))*100
```

```python
# calculate f1 score
f1_score(y_test, lr.predict(x_test))*100
```
