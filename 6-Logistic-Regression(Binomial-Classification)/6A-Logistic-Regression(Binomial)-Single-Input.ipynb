## Logistic Regression - Binomial Classification (Single Input)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
dataset.head(3)
```

```python
df = dataset[['Age', 'Purchased']]
df.head(5)
```

```python
# Visualize the data to check whether it is applicable for logistic regression or not
plt.figure(figsize=(6,4))
sns.scatterplot(x='Age', y='Purchased', data=df)
plt.show()
```

```python
# check data types and null values
df.info()
```

```python
# Create input and Output variables
x = df[['Age']] # Input must be 2D
y = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
```

```python
# split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Initialize and train logistic regression
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
# Accuracy
lr.score(x_test, y_test)*100
```

```python
# pedict Purchase for age 48
lr.predict([[48]]) # output will be 0/1
```

```python
y_pred = lr.predict(x)
```

```python
# Visualize the logistic regression plot (sigmoid function graph)
plt.figure(figsize=(6,4))
sns.scatterplot(x='Age', y='Purchased', data=df)
sns.lineplot(x='Age', y= y_pred , data=df, color='green')
plt.title('Logistic Regression (Binomial)')
plt.show()
```
