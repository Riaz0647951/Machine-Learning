### Imbalanced Data Handling Techniques (Random Under Sampling & Random Over Sampling)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head(3)
```

```python
df = df.drop(columns=['User ID', 'Gender'])
df.head(3)
```

```python
df['Purchased'].value_counts()
```

The output variable (Purchased) has two classes (0,1) but it is imbalanced (0 ---> 257) and (1 ---> 143). 

So first, we will use Random Under Sampling Technique and then Random Over Sampling.

```python
x = df.iloc[:,:-1]
y = df['Purchased']
```

#### Random Under Sampling

It is used to reduce the majority section of classes so that it matches the minority number of classes

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
rus = RandomUnderSampler()
rus_x, rus_y = rus.fit_resample(x,y)
```

```python
rus_y.value_counts()
```

**Now the dataset is balanced using "Random Under Sampling" method as it has 143 values as 0 and 143 values as 1**

#### Random Over Sampling

It is used to increase the minority section of classes so that it matches the majority number of classes

```python
from imblearn.over_sampling import RandomOverSampler
```

```python
ros = RandomOverSampler()
ros_x, ros_y = ros.fit_resample(x, y)
```

```python
ros_y.value_counts()
```

**The dataset is balanced using "Random Over Sampling" method as it has 257 values as 0 and 257 values as 1**
